## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision improved: which content pages a review team should look at first, out of thousands, given limited reviewer time. Action: a content strategist opens the top-ranked pages and decides whether to refresh, expand, or leave them. Cost of a wrong call: if the ranking is wrong, reviewers waste limited hours on pages that were never actually declining (false positive), or a genuinely declining high-traffic page gets missed and keeps losing visibility for weeks before anyone notices (false negative) — the second failure is more expensive, since it compounds silently.
